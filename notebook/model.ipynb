## **BUSINESS UNDERSTANDING**

### **Konteks Utama:** 
Proyek AI Learning Insight ini memanfaatkan dataset learning journey dari database Dicoding, yang mencakup riwayat aktivitas belajar, pengumpulan tugas (submissions), hingga nilai ujian siswa di berbagai kelas.

### **1. Latar Belakang Masalah**
Saat ini, platform Dicoding melayani ribuan siswa dengan latar belakang dan kebiasaan belajar yang sangat beragam. Namun, pendekatan pembelajaran dan intervensi yang diberikan masih cenderung seragam (one-size-fits-all).<br>

Tanpa pemahaman mendalam tentang pola perilaku belajar siswa, platform berpotensi kehilangan kesempatan strategis untuk:
- Mendeteksi karakteristik dan kebiasaan belajar aktual siswa selama mengikuti kelas.
- Memberikan intervensi atau pendampingan yang spesifik sesuai dengan hambatan nyata yang dialami siswa.
- Mencegah risiko gagal atau drop-out secara dini sebelum siswa benar-benar berhenti belajar.

**Oleh karena itu, diperlukan segmentasi berbasis data (data-driven) untuk menemukan kelompok-kelompok profil belajar. Hasil segmentasi ini nantinya akan dianalisis untuk mendefinisikan label Learning Behavior Type dari masing-masing kelompok sebagai dasar pemberian rekomendasi yang terpersonalisasi.**

### **2. Pertanyaan Bisnis Utama**
Proyek ini bertujuan menjawab pertanyaan strategis berikut:

1. **Berapa banyak klaster perilaku belajar optimal yang secara natural terbentuk dari data interaksi siswa di Dicoding?**


2. **Apa karakteristik dominan dari setiap klaster yang nantinya akan mendefinisikan label Learning Behavior Type mereka?** (Misal: metrik kecepatan, konsistensi, dan ketangguhan apa yang membedakan satu klaster dengan klaster lainnya?)

3. **Bagaimana justifikasi parameter untuk setiap tipe perilaku belajar tersebut**(Indikator spesifik apa yang memvalidasi alasan seorang siswa dikategorikan ke dalam tipe tertentu?)

4. **Apa rekomendasi intervensi dan perbaikan yang paling tepat untuk masing-masing Learning Behavior Type?** (Strategi pendampingan, manajemen waktu, atau pengingat spesifik seperti apa yang paling dibutuhkan oleh tiap kelompok?)

### **3. Tujuan Proyek**
Proyek ini mengimplementasikan arsitektur yang menggabungkan Machine Learning (Clustering) dan Generative AI (LLM) dengan rincian tujuan sebagai berikut:

- **Tahap 1 - Clustering (Machine Learning):** Membentuk klaster-klaster terbaik secara objektif dari data mentah berdasarkan metrik historis perilaku belajar siswa.

- **Tahap 2 - Analisis Karakteristik (Analytical):** Mengekstrak parameter utama (centroid/highlight parameters) dari setiap klaster yang terbentuk untuk mendefinisikan dan memberikan label Learning Behavior Type.

- **Tahap 3 - Interpretasi & Rekomendasi (Generative AI / LLM):** Menggunakan Large Language Model untuk membaca karakteristik klaster, menjustifikasi penempatan siswa berdasarkan parameter dominannya, dan merumuskan rekomendasi intervensi yang actionable bagi tim bisnis Dicoding.

*(Catatan: Notebook ini akan berfokus pada eksekusi Tahap 1 dan 2, yang akan menghasilkan landasan data yang esensial untuk dieksekusi oleh sistem LLM di Tahap 3).*

### **4. Dimensi Analisis (Fokus Penilaian)**
Untuk membentuk klaster yang berkualitas, perilaku siswa dievaluasi berdasarkan 4 dimensi utama:

1. **Kecepatan & Efisiensi:** Seberapa cepat siswa menyelesaikan materi dibandingkan dengan estimasi waktu standar platform.

2. **Konsistensi & Disiplin:** Seberapa rutin siswa mengakses materi dan mempertahankan ritme belajarnya.

3. **Kualitas Usaha & Iterasi:** Seberapa baik nilai tugas yang dikumpulkan dan tingkat intensitas/kegigihan siswa dalam melakukan revisi tugas.

4. **Ketangguhan Akademik (Resilience):** Tingkat keberhasilan kelulusan ujian dan kegigihan siswa dalam mengulang ujian (remedial) jika mengalami kegagalan.

### **5. Output yang Diharapkan (Deliverables)**
1. **Model Clustering Terbaik:** Algoritma machine learning teruji yang mampu memisahkan kelompok siswa dengan batasan profil yang jelas dan metrik evaluasi yang valid (tersimpan dalam format model yang dapat di-export dan di-load untuk memprediksi klaster siswa baru).

2. **Profil Learning Behavior Type:** Definisi detail dari karakteristik tiap klaster berdasarkan nilai parameter dominannya (centroids).

3. **Konteks Rekomendasi Berbasis LLM:** Rancangan prompt dan data terstruktur yang siap diumpankan ke sistem LLM untuk men-generate strategi penanganan otomatis (desain reminder, strategi belajar) bagi tim operasional Dicoding.

### **6. Kriteria Keberhasilan (Success Criteria)**
**Dari Sisi Teknis & Analitik:**

- Model clustering mencapai evaluasi metrik yang memadai (misal: Silhouette Score yang merepresentasikan pemisahan klaster yang optimal).

- Tidak ada klaster noise yang terlalu kecil proporsinya (misal < 5% populasi) atau klaster mayoritas yang mendominasi secara tidak wajar.

**Dari Sisi Bisnis:**

- Analisis karakteristik klaster berhasil menerjemahkan angka metrik menjadi profil Learning Behavior Type yang logis, berbeda secara signifikan antar klaster, dan bebas tumpang tindih.

- Parameter yang dihasilkan dapat dicerna dengan baik oleh LLM untuk memproduksi justifikasi dan rekomendasi intervensi yang relevan dan dapat dieksekusi secara nyata oleh platform.

### **7. Pihak yang Berkepentingan (Stakeholders & Use Cases)**
Untuk memastikan insight yang dihasilkan memberikan nilai bisnis yang nyata, hasil dari proyek ini ditujukan untuk beberapa stakeholder utama:

- **Tim Produk (Product Manager):** Menggunakan profil klaster untuk merancang fitur personalisasi alur belajar (learning path) dan penyesuaian UX platform.

- **Tim Konten (Course Reviewer / Creator):** Mendapatkan pemahaman tentang letak hambatan (bottleneck) pembelajaran pada segmen siswa tertentu, sehingga tingkat kesulitan materi dapat disesuaikan.

### **8. Batasan Proyek (Project Scope)**
Agar proyek tetap fokus pada tujuan penyelesaian masalah utama, batasan ruang lingkup (out-of-scope) didefinisikan sebagai berikut:

## **MERGING**

````python
# memastikan kolom waktu dalam format datetime
time_cols_tracking = ['first_opened_at', 'completed_at']
time_cols_submission = ['created_at', 'updated_at']
time_cols_exam_reg = ['created_at', 'updated_at', 'deadline_at', 'retake_limit_at', 'exam_finished_at', 'deleted_at', 'deadline_at']
time_cols_completion = ['created_at', 'updated_at']

def convert_to_datetime(df, cols):
    for col in cols:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')
    return df

df_tracking = convert_to_datetime(df_tracking_fs, time_cols_tracking)
df_submission = convert_to_datetime(df_submission_fs, time_cols_submission)
df_exam_reg = convert_to_datetime(df_exregis_fs, time_cols_exam_reg)
df_completion = convert_to_datetime(df_complation_fs, time_cols_completion)
df_users = convert_to_datetime(df_user_fs, ['created_at'])
````

````python
# A. MASTER ACTIVITY TABLE
master_activity = df_tracking.merge(
    df_tutorial_fs[['id', 'position', 'type']],
    left_on='tutorial_id',
    right_on='id',
    how='left',
    suffixes=('', '_tut')
).drop(columns=['id_tut'])

master_activity = master_activity.merge(
    df_journey_fs[['id', 'difficulty', 'hours_to_study', 'type']],
    left_on='journey_id',
    right_on='id',
    how='left',
    suffixes=('', '_jour')
).drop(columns=['id_jour'])

master_activity.head()
````

💡 **Explanation:**<br>
**Pembuatan Master Activity Table**<br>
Kode ini menyatukan log aktivitas belajar dengan informasi modul dan kelas untuk membentuk satu dataset utama yang utuh.

- **Merge 1** (df_tutorial_fs): Melampirkan detail modul (urutan/position dan type) ke data aktivitas berdasarkan tutorial_id.
- **Merge 2** (df_journey_fs): Melampirkan detail kelas (difficulty, estimasi waktu, dan tipe kelas) berdasarkan journey_id.
- **Pembersihan** (drop): Membuang kolom ID duplikat (id_tut, id_jour) hasil penggabungan agar tabel tidak redundan.

**Tujuan Akhir:** Menyiapkan satu tabel sumber yang kaya konteks (Single Source of Truth) untuk diekstrak menjadi fitur-fitur pemodelan clustering nantinya.

````python
# B. MASTER PROJECT TABLE (Submission + Journey Info)
master_project = df_submission.merge(
    df_journey_fs[['id', 'difficulty', 'type', 'deadline']],
    left_on='journey_id',
    right_on='id',
    how='left',
    suffixes=('', '_jour')
).drop(columns=['id_jour'])

master_project.head()
````

💡 **Explanation:**<br>
**Pembuatan Master Project Table**<br>
Kode ini menggabungkan riwayat pengumpulan tugas (submission) dengan informasi detail dari kelas/journey yang bersangkutan.

- **Merge** (df_journey_fs): Melampirkan metrik kelas (tingkat kesulitan/difficulty, jenis/type, dan tenggat waktu/deadline) ke dalam data submission berdasarkan pencocokan journey_id.
- **Pembersihan** (drop): Membuang kolom ID duplikat (id_jour) hasil penggabungan agar struktur tabel tetap bersih.

**Tujuan Akhir:** Membentuk dataset komprehensif terkait performa tugas siswa, yang nantinya digunakan untuk mengekstrak fitur seperti kepatuhan pada deadline atau kualitas submission berdasarkan tingkat kesulitan kelas.

````python
#  C. MASTER EXAM TABLE (Result + Registration + Tutorial + journey)
df_exresult_fs_clean = df_exresult_fs.drop_duplicates(subset=['exam_registration_id'], keep='last')

master_exam = df_exresult_fs_clean.merge(
    df_exam_reg,
    left_on='exam_registration_id',
    right_on='id',
    how='right',
    suffixes=('_res', '_reg')
)

master_exam = master_exam.merge(
    df_tutorial_fs[['id', 'developer_journey_id', 'position']],
    left_on='tutorial_id',
    right_on='id',
    how='left'
).rename(columns={'id': 'tutorial_id_ref'})

master_exam = master_exam.merge(
    df_journey_fs[['id', 'difficulty', 'hours_to_study']],
    left_on='developer_journey_id',
    right_on='id',
    how='left'
)

master_exam.head()
````

💡 **Explanation:**<br>
**Pembuatan Master Exam Table**<br>
Kode ini menyatukan riwayat pendaftaran ujian dan hasil ujian dengan konteks modul serta detail kelasnya.

- **Merge 1** (df_exam_reg): Menggabungkan hasil ujian dengan data registrasi secara utuh (Outer Join) untuk melacak status ujian beserta batas waktunya.
- **Merge 2** (df_tutorial_fs): Melampirkan informasi modul (posisi modul dan developer_journey_id) berdasarkan pencocokan tutorial_id. Kolom ID tambahan diubah namanya menjadi tutorial_id_ref agar tidak bingung.
- **Merge 3** (df_journey_fs): Melampirkan detail tingkat kelas (difficulty dan estimasi waktu) berdasarkan developer_journey_id dari hasil merge sebelumnya.

**Tujuan Akhir:** Membentuk dataset komprehensif terkait performa ujian siswa, yang akan digunakan untuk mengekstrak fitur ketangguhan akademik (seperti nilai rata-rata ujian, rasio kelulusan, dan frekuensi ujian ulang/remedial).

## **FEATURE ENGINEERING**

````python
# 1. mapping bobot kesulitan (Expert > Beginner)
DIFFICULTY_MAP = {
    'beginner': 1,
    'intermediate': 2,
    'expert': 3,
    'advanced': 3
}

def get_difficulty_weight(val):
    return DIFFICULTY_MAP.get(str(val).lower(), 1)

def calculate_streak(dates):
    dates = pd.Series(dates).dropna()
    
    if len(dates) == 0: return 0

    dates = sorted(list(set(dates)))
    
    longest_streak = 0
    current_streak = 0
    prev_date = None

    for d in dates:
        if prev_date is None:
            current_streak = 1
        elif (d - prev_date).days == 1:
            current_streak += 1
        else: 
            longest_streak = max(longest_streak, current_streak)
            current_streak = 1
        prev_date = d
            
    return max(longest_streak, current_streak)
````

### **FEATURE CREATION**

````python
# A. STREAM AKTIVITAS (SPEED & TIME)
valid_activity = master_activity[
    (master_activity['status'] == 1) & 
    (master_activity['first_opened_at'].notnull()) & 
    (master_activity['completed_at'].notnull())
].copy()

valid_activity['duration_hours'] = (valid_activity['completed_at'] - valid_activity['first_opened_at']).dt.total_seconds() / 3600
valid_activity = valid_activity[valid_activity['duration_hours'] > 0.01] # Hapus glitch 0 detik

# rumus: durasi aktual / estimasi durasi dari journey
valid_activity['hours_to_study'] = valid_activity['hours_to_study'].replace(0, 1)
valid_activity['speed_ratio'] = valid_activity['duration_hours'] / valid_activity['hours_to_study']
valid_activity['speed_ratio'] = valid_activity['speed_ratio'].clip(upper=5.0) # selambat-lambatnya orang, anggap maksimal 5x lipat dari waktu normal

master_activity['date'] = master_activity['first_opened_at'].dt.date
master_activity['hour'] = master_activity['first_opened_at'].dt.hour
master_activity['is_weekend'] = master_activity['first_opened_at'].dt.dayofweek.isin([5, 6]).astype(int)


# B. STREAM PROJECT (PROCRASTINATION & QUALITY)
journey_start_dates = master_activity.groupby(['developer_id', 'journey_id'])['first_opened_at'].min().reset_index()
journey_start_dates.rename(columns={'first_opened_at': 'start_date', 'developer_id': 'submitter_id'}, inplace=True)

master_project = master_project.merge(
    journey_start_dates,
    on=['submitter_id', 'journey_id'],
    how='left'
)

master_project['weight'] = master_project['difficulty'].apply(get_difficulty_weight)

# deadline = integer (misal 135 hari)
# rumus: (start date + deadline days) - submission date
master_project['start_date'] = pd.to_datetime(master_project['start_date'])
master_project['created_at'] = pd.to_datetime(master_project['created_at'])
master_project['deadline'] = pd.to_numeric(master_project['deadline'], errors='coerce').fillna(30) # Default 30 hari jika null

master_project['days_taken'] = (master_project['created_at'] - master_project['start_date']).dt.total_seconds() / (24*3600)

# positif = masih punya sisa waktu (aman)
# negatif = lewat deadline (telat)
# keterlambatan maksimal yang dihitung model adalah 30 hari (-30), dan pengumpulan paling awal adalah 30 hari sebelum deadline
# mendekati 0 = deadliner
master_project['days_before_deadline'] = master_project['deadline'] - master_project['days_taken']
master_project['days_before_deadline'] = master_project['days_before_deadline'].clip(lower=-30, upper=30)
master_project['days_before_deadline'] = master_project['days_before_deadline'].fillna(0)


# C. STREAM EXAM (PERFORMANCE & GRIT)
master_exam['is_retake'] = master_exam['deleted_at'].notnull().astype(int)
master_exam['weight'] = master_exam['difficulty'].apply(get_difficulty_weight)
````

````python
# A. AGGREGASI AKTIVITAS (BEHAVIOR)
user_activity_agg = master_activity.groupby('developer_id').agg({
    'date': ['nunique', calculate_streak],    
    'is_weekend': 'mean',                      
    'id': 'count',                          
    'tutorial_id': 'nunique'                
})

user_activity_agg.columns = ['total_active_days', 'longest_streak', 'weekend_ratio', 'total_clicks', 'unique_modules']

avg_speed = valid_activity.groupby('developer_id')['speed_ratio'].mean().rename('avg_speed_ratio')
user_activity_agg = user_activity_agg.join(avg_speed)

user_activity_agg['review_rate'] = (user_activity_agg['total_clicks'] - user_activity_agg['unique_modules']) / user_activity_agg['total_clicks']

user_preferred_time = master_activity.groupby('developer_id')['hour'].agg(
    lambda x: x.mode()[0] if not x.mode().empty else 12
).rename('preferred_study_hour')


# B. AGGREGASI PROJECT (QUALITY & PROCRASTINATION) ---
user_project_agg = master_project.groupby('submitter_id').agg({
    'rating': 'mean',                          
    'weight': 'mean',                          
    'id': ['count'],                           
    'days_before_deadline': 'mean',          
    'quiz_id': 'nunique',   
})

user_project_agg.columns = [
    'avg_project_score', 
    'avg_project_difficulty', 
    'total_submissions_count',  
    'avg_procrastination_days', 
    'total_projects_completed'  
]

# rumus: (total submit / total project unik)
# contoh: 5 kali submit untuk 1 project = 5 attempt (gigih)
user_project_agg['avg_attempts_per_project'] = (
    user_project_agg['total_submissions_count'] / user_project_agg['total_projects_completed']
).fillna(0)

# C. AGGREGASI EXAM (PERFORMANCE & GRIT)
user_exam_agg = master_exam.groupby('examinees_id').agg({
    'score': 'mean',                           
    'is_passed': 'mean',                       
    'is_retake': 'sum',                        
    'weight': 'mean'                           
}).rename(columns={
    'score': 'avg_exam_score',
    'is_passed': 'exam_pass_rate',
    'is_retake': 'total_exam_retakes',
    'weight': 'avg_exam_difficulty'
})
````

````python
# cek distribusi skewness
def check_skewness(df, features, threshold=0.5):
    results = []

    for col in features:
        skewness_value = skew(df[col].dropna())
        
        if skewness_value < -threshold:
            skew_class = 'Negatively Skewed'
        elif skewness_value > threshold:
            skew_class = 'Positively Skewed'
        else:
            skew_class = 'Symmetric'
        
        results.append({
            'Feature': col,
            'Skewness': skewness_value,
            'Skew_Type': skew_class
        })

    skew_df = pd.DataFrame(results)
    return skew_df

skew_df = check_skewness(final_df_clean, features)
print(skew_df)
````

````python
def plot_silhouette(df_scaled, cluster_labels, optimal_k, sil_score):
    sample_silhouette_values = silhouette_samples(df_scaled, cluster_labels)

    fig, ax1 = plt.subplots(figsize=(10, 7))

    # set axis limits
    ax1.set_xlim([-0.2, 1.0])
    ax1.set_ylim([0, len(df_scaled) + (optimal_k + 1) * 10])

    y_lower = 10

    for i in range(optimal_k):
        # get silhouette values for each cluster
        ith_cluster_silhouette_values = sample_silhouette_values[cluster_labels == i]
        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / optimal_k)

        ax1.fill_betweenx(
            np.arange(y_lower, y_upper),
            0,
            ith_cluster_silhouette_values,
            facecolor=color,
            edgecolor=color,
            alpha=0.7
        )

        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        y_lower = y_upper + 10

    ax1.set_title(f"silhouette plot for k-means (k={optimal_k})", fontsize=16)
    ax1.set_xlabel("silhouette coefficient", fontsize=12)
    ax1.set_ylabel("cluster label", fontsize=12)

    # draw average silhouette line
    ax1.axvline(x=sil_score, color="red", linestyle="--", label=f"average: {sil_score:.2f}")

    ax1.set_yticks([])
    ax1.legend(loc="upper right")

    plt.tight_layout()
    plt.show()

plot_silhouette(df_scaled, cluster_labels, OPTIMAL_K, sil_score)
````

````python
def plot_tsne_clusters(final_df, cluster_labels, perplexity=10, random_state=42):
    from sklearn.manifold import TSNE
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns

    # prepare feature matrix
    X_viz = final_df.drop(columns=['cluster_label'], errors='ignore')

    print("processing t-sne...")
    tsne = TSNE(
        n_components=2,
        perplexity=perplexity,
        max_iter=1000,
        verbose=1,
        random_state=random_state
    )

    tsne_results = tsne.fit_transform(X_viz)

    df_tsne = pd.DataFrame(tsne_results, columns=["Dim1", "Dim2"])
    df_tsne["Cluster"] = cluster_labels

    # compute visual centroids
    visual_centroids = (
        df_tsne.groupby("Cluster")[["Dim1", "Dim2"]]
        .mean()
        .reset_index()
    )

    plt.figure(figsize=(12, 8))

    # scatter plot of samples
    sns.scatterplot(
        data=df_tsne,
        x="Dim1",
        y="Dim2",
        hue="Cluster",
        palette="bright",
        alpha=0.6,
        s=60,
        legend="full"
    )

    # plot centroid markers
    plt.scatter(
        visual_centroids["Dim1"],
        visual_centroids["Dim2"],
        c="black",
        s=300,
        marker="X",
        label="Centroid",
        edgecolor="white",
        linewidth=1.5
    )

    # label centroid with cluster id
    for _, row in visual_centroids.iterrows():
        plt.annotate(
            str(int(row["Cluster"])),
            (row["Dim1"], row["Dim2"]),
            xytext=(0, 20),
            textcoords="offset points",
            ha="center",
            fontsize=14,
            weight="bold",
            bbox=dict(boxstyle="round,pad=0.3", fc="white", alpha=0.8)
        )

    plt.title("cluster visualization (t-sne)", fontsize=16)
    plt.xlabel("")
    plt.ylabel("")
    plt.legend(title="cluster id", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

plot_tsne_clusters(df_scaled, cluster_labels)
````

## **HYPERPARAMETER TUNNING**

````python
# inisialisasi parameter
param_grid = {
    'n_clusters': [2, 3, 4, 5],        
    'init': ['k-means++', 'random'],   
    'n_init': [10, 20, 50],            
    'max_iter': [300, 500]            
}

# membuat kemungkinan kombinasi parameter
keys = param_grid.keys()
values = (param_grid[key] for key in keys)
combinations = [dict(zip(keys, combination)) for combination in itertools.product(*values)]

# simpan hasil terbaik
best_silhouette = -1
best_params = None
best_model = None
best_dbi = float('inf')
best_labels = None
````

````python
for params in combinations:
    # inisialisasi model dengan kombinasi
    km = KMeans(**params, random_state=42)
    
    # fit & predict
    labels = km.fit_predict(df_scaled)
    
    # hitung metrik evaluasi
    sil_score = silhouette_score(df_scaled, labels)
    dbi_score = davies_bouldin_score(df_scaled, labels)
    
    if sil_score > best_silhouette:
        best_silhouette = sil_score
        best_params = params
        best_model = km
        best_dbi = dbi_score
        best_labels = labels
````

````python
print("========== HASIL TUNNING TERBAIK ==========")
print(f"Best Silhouette Score     : {best_silhouette:.4f} (Mendekati 1 = Baik)")
print(f"Best Davies-Bouldin Index : {best_dbi:.4f} (Mendekati 0 = Baik)")
print(f"Best Parameters           : {best_params}")
````
